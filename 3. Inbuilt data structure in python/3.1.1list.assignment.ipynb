## Assignment 9: Matrix Transposition

Write a function that takes a 3x3 matrix (nested list) as input and returns its transpose. Print the original and transposed matrices.

```python
# Function to transpose a 3x3 matrix
def transpose_matrix(matrix):
    # Using list comprehension to transpose
    transposed = [[matrix[j][i] for j in range(3)] for i in range(3)]
    return transposed

# Original 3x3 matrix
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Get the transposed matrix
transposed_matrix = transpose_matrix(matrix)

# Print both matrices
print("Original Matrix:")
for row in matrix:
    print(row)

print("\nTransposed Matrix:")
for row in transposed_matrix:
    print(row)
```



### Assignment 10: Flattening a Nested List

Write a function that takes a nested list and flattens it into a single list. Print the original 

```python
# Function to flatten a nested list
def flatten_list(nested_list):
    flat_list = []          # create an empty list to store all items
    for sublist in nested_list:   # go through each small list inside the big list
        for item in sublist:      # go through each element inside the small list
            flat_list.append(item)  # add each element to the flat list
    return flat_list

# Original nested list
nested_list = [[1, 2, 3], [4, 5], [6, 7, 8]]

# Call the function to flatten the list
flattened = flatten_list(nested_list)

# Print the results
print("Original Nested List:")
print(nested_list)

print("\nFlattened List:")
print(flattened)
```


### Assignment 13: List Reversal

Write a function that takes a list and returns a new list with the elements in reverse order. Print the original and reversed lists.


```python
# Function to reverse a list
def reverse_list(lst):
    return lst[::-1]   # slicing method to reverse the list

# Original list
numbers = [1, 2, 3, 4, 5, 6]

# Call the function to get reversed list
reversed_numbers = reverse_list(numbers)

# Print both lists
print("Original List:")
print(numbers)

print("\nReversed List:")
print(reversed_numbers)
```

### Assignment 14: List Rotation

Write a function that rotates a list by n positions. Print the original and rotated lists.


```python
# Function to rotate a list by n positions
def rotate_list(lst, n):
    n = n % len(lst)   # handle if n is larger than list length
    return lst[-n:] + lst[:-n]  # move last n elements to the front

# Original list
numbers = [1, 2, 3, 4, 5]

# Rotate by 2 positions
rotated_numbers = rotate_list(numbers, 2)

# Print the results
print("Original List:")
print(numbers)

print("\nRotated List (by 2 positions):")
print(rotated_numbers)
```

### Assignment 15: List Intersection

Write a function that takes two lists and returns a new list containing only the elements that are present in both lists. Print the intersected list.

```python
# Function to find intersection of two lists
def list_intersection(list1, list2):
    result = []
    for item in list1:
        if item in list2 and item not in result:  # avoid duplicates
            result.append(item)
    return result

# Example lists
list_a = [1, 2, 3, 4, 5]
list_b = [3, 4, 5, 6, 7]

# Find intersection
intersection = list_intersection(list_a, list_b)

# Print the result
print("List A:", list_a)
print("List B:", list_b)
print("\nIntersection:")
print(intersection)
```
