
### Assignment 3: Set Operations

Create two sets: one with the first 5 positive integers and another with the first 5 even integers. Perform and print the results of union, intersection, difference, and symmetric difference operations on these sets.

```python
five_positive_intiger={1,2,3,4,5}
five_even_integer={2,4,6,8,10}


uniun=five_even_integer.union(five_positive_intiger)
print(uniun)

intersection=five_positive_intiger.intersection(five_even_integer)
print(intersection)

symmetric_difference=five_positive_intiger.symmetric_difference(five_even_integer)
print(symmetric_difference)

```
