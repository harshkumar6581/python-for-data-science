### Accessing dictionary item in python  
- accessing value using key in names
-  accessing value using get () method
-  accessing value using keys () method
- accessing value using value method




````python
#  accessing value using key in names
car = {'brand' :'audi','model':'q7'}## audi is the key value 
print(car['brand'])

print(car['model'])

print(car['brand'], car['model'])
````

````python
# accessing value using get () method
car = {'brand' :'audi','model':'q7'}
print(car.get('brand'))
````

````python
# accessing value using keys () method 
car = {'brand' :'audi','model':'q7'}
car_keys = car.keys()
print(car_keys)
car['fuel type'] = 'diesel'
car_keys
````

````python
# accessing value using value method
car = {'brand' :'audi','model':'q7'}
car_values = car.values()
print(car_values) 
````

````python
## accessing item using item () method

car = {'brand' :'audi','model':'q7'}
car_item = car.items()
print(car.items)
car['fuel type']= 'diesel'
print(car_item)
````
