# Python Operators

## 1. Introduction to Operators
Operators are special symbols in Python that perform operations on variables and values.  
Example: `5 + 3` → `+` is the operator.

---

## 2. Arithmetic Operators
Used for mathematical calculations.  

- `+` → Addition  
- `-` → Subtraction  
- `*` → Multiplication  
- `/` → Division (always gives float)  
- `%` → Modulus (remainder)  
- `//` → Floor Division (integer result)  
- `**` → Exponentiation (power)  


# Python Operators (Detailed Notes)

## 3. Assignment Operators
Used to give values to variables or update them.

- `=` → Simple assignment  
- `+=` → Add then assign  
- `-=` → Subtract then assign  
- `*=` → Multiply then assign  
- `/=` → Divide then assign  

**Example:**
```python
x = 10      # assign
x += 5      # x = x + 5 → 15
x -= 3      # x = x - 3 → 12
x *= 2      # x = x * 2 → 24
x /= 4      # x = x / 4 → 6.0
print(x)

## 4. Comparison / Relational Operators
Used to compare two values. The result is either `True` or `False`.

- `==` → Equal to  
- `!=` → Not equal to  
- `>`  → Greater than  
- `<`  → Less than  
- `>=` → Greater than or equal to  
- `<=` → Less than or equal to  

**Example:**
```python
a = 10
b = 7

print(a == b)   # False
print(a != b)   # True
print(a > b)    # True
print(a < b)    # False
print(a >= b)   # True
print(a <= b)   # False

## 5. Logical Operators
Logical operators deal with `True`/`False` values. They are useful when combining conditions.

- `and` → True if **both** conditions are true  
- `or`  → True if **at least one** condition is true  
- `not` → Reverses the result (True → False, False → True)  

**Example:**
```python
age = 20
has_id = False

print(age > 18 and has_id)   # False
print(age > 18 or has_id)    # True
print(not has_id)            # True
