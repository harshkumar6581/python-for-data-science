What is a Variable?

A variable is like a container (box) where we store data (a value).
We give the box a name, so we can use it later in our program.

A variable is a name that is used to store a value in memory, so we can use that value later in the program.

### Type cheking and conversion 
. Typecasting is the process of converting one data type to another.
. python provide buildes built-in function for type casting


             .int():Convert to integer.
             .float():Convert to float.
             .str():Convert to str.
             .bool():Convert to bool.


```python
a = 34
b = "34"
d = 3322

print(a)
print(type(a))

print(b)
print(type(b))

# convert b to an integer
c = int(b)
print(c)
print(type(c))

e = str(d)
print(e)
print(type(e))
```
