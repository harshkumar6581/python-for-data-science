### 📌 Project Description

This is a simple Python Rock-Paper-Scissors game that runs in the terminal. The program uses Python’s built-in random module to generate the computer’s choice and takes user input to play against it. It compares both choices using conditional logic to determine whether the result is a win, loss, or tie, and then displays the result on the screen. This project demonstrates basic Python concepts such as functions, lists, user input, conditional statements, and importing modules.

```python
import random
def game_play():
    option=["rock","paper","scissor"]
    comp_choice=random.choice(option)
    user_choice=input("choose Rock,paper,or scissor:").lower()
    print(f"\n you chose: {user_choice}")
    print(f"\n computer chose: {comp_choice}")
    if user_choice==comp_choice:
        print("tie")
    elif user_choice=="rock" and comp_choice=="scissor":
        print("you will win ")
    elif user_choice=="paper" and comp_choice=="rock":
        print("you will win ")
    elif user_choice=="scissors" and comp_choice=="paper":
        print("you will win ")   
    else:
        print(f"you lose ") 
game_play()                
```
