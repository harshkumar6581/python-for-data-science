### importing module in python: Modules and packages 

in python, module and packages helps organize and reused code.here's comprehensive guide how to iport them.

```python
import math#math is a packages 
math.sqrt(9)
```

```python
math.sqrt(16)
```

```python
from math import sqrt,pi
print(math.sqrt(25))
print(math.sqrt(16))
print(pi)
```
