#### polymorphism 
 Polymorphism is a core concept in Object oriented programing (oop) that allow object of different classes to be treated as object of a common superclass. it provides a way to perform to perform a single action in different form Polymorphism is typicallay achive through method overriding and interface

### Method Overriding 
Method overriding allows a child class to provie a specific implimentation of  a methosd that is allready define in the parent class 


```python
## base class 
class Animal:
    def speak (self):
        return "sount of the animal "

## derive class 1
class Dog(Animal):
    def speak(self):
        return "woof"

## Derive class 
class Cat(Animal):
    def speak(self):
        return"meow"

## function that demonstratte polymorphism 
def animal_speak(animal):
    print(animal.speak())        

dog=Dog()
cat=Cat()
print(dog.speak())
print(cat.speak())       
```

```python
## polymorphism with function and method 
## base class

class Shape:
    def area(self):
        return "area of the figure "


## derive class 
class Rectangle(Shape):
    def __init__(self,width,hight):
        self.width= width
        self.hight= hight


    def area(self):
        return self.width * self.hight


# derive class 2 
class Circle(Shape):
    def __init__(self,radious):
        self.radious=radious
    
    def area(self):
        return 3.14*self.radious*self.radious

## function that demonstrate polymorphism 
def print_area(Shape):
    print(f"the area is {Shape.area()}") 

rectangle=Rectangle(4,5)
circle=Circle(3)

print_area(rectangle)
print_area(circle)

```

### polymorphism with abstract base class 
Abstract Base Class (ABCs) are used to define common method for group of related objects.They can enforce that derive class impliment particular method , promot consistency across different implimentation 

```python
from abc import ABC,abstractmethod 
## define an abstract class 
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass


## Derive class1 
class Car(Vehicle):
    def start_engine(self):
        return "car engine started"


## Derive class 2 
class Motorcycle(Vehicle):
    def start_engine(self):
        return "motorcyce engine startd"

# Funtion that demonstrate polymorphism 

def start_vehicle(Vehicle):
    print(Vehicle.start_engine())


## create object of car and motorcycle 
car= Car()
motorcycle =Motorcycle()
start_vehicle(car)
```
