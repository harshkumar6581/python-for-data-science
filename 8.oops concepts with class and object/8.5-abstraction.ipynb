### Abstraction
 Abstraction is the concept of hiding the complex thing and shoe you the essential necessary features of an object. This help in reducing the programing complexity and efforts.

```python
from abc import ABC,abstractmethod

## abstract class]
class Vehicle(ABC):
    def drive(self):
        print("the vehicle is used for driving ")

    @abstractmethod
    def star_engine(self):
        pass    
 
class Car(Vehicle):
    def star_engine(self):
        print("car engine started")


def operate_vehicle(vehicle):
    vehicle.star_engine()
    vehicle.drive()


car=Car()

operate_vehicle(car)
```
