### Creating own data type in python 
This project creates a custom Fraction data type in Python using object-oriented programming. It supports addition, subtraction, multiplication, division, and decimal conversion using operator overloading. This project demonstrates magic methods and custom class behavior.


Features

- Custom Fraction class

- Operator overloading

- Decimal conversion

- Error handling for zero denominator

```python
class fraction:
    ## parameterized constructor
    def __init__(self,x,y):
        self.num=x
        self.den=y

        
    def __str__(self):
        return '{}/{}'.format(self.num,self.den)


    def __add__(self,other):
        new_num=self.num*other.den+other.num*self.den
        new_den=self.den*other.den
        return '{}/{}'.format(new_num,new_den)
    def __sub__(self,other):
        new_num=self.num*other.den - other.num*self.den
        new_den=self.den*other.den
        return '{}/{}'.format(new_num,new_den)
    def __mul__(self,other):
        new_num=self.num*other.den*other.num*self.den
        new_den=self.den*other.den
        return '{}/{}'.format(new_num,new_den)
    def __truediv__(self,other):
        new_num=self.num*other.den/other.num*self.den
        new_den=self.den*other.den
        return '{}/{}'.format(new_num,new_den)

    def convert_to_decimal(self):
        return self.num/self.den
```

```python
class Fraction:
    """
    A custom Fraction data type that supports
    addition, subtraction, multiplication,
    division and decimal conversion.
    """

    # Parameterized constructor
    def __init__(self, numerator, denominator):
        if denominator == 0:
            raise ValueError("Denominator cannot be zero")
        self.num = numerator
        self.den = denominator

    # String representation
    def __str__(self):
        return f"{self.num}/{self.den}"

    # Addition
    def __add__(self, other):
        new_num = self.num * other.den + other.num * self.den
        new_den = self.den * other.den
        return Fraction(new_num, new_den)

    # Subtraction
    def __sub__(self, other):
        new_num = self.num * other.den - other.num * self.den
        new_den = self.den * other.den
        return Fraction(new_num, new_den)

    # Multiplication
    def __mul__(self, other):
        new_num = self.num * other.num
        new_den = self.den * other.den
        return Fraction(new_num, new_den)

    # Division
    def __truediv__(self, other):
        new_num = self.num * other.den
        new_den = self.den * other.num
        return Fraction(new_num, new_den)

    # Convert to decimal
    def convert_to_decimal(self):
        return self.num / self.den


# Example Usage
if __name__ == "__main__":
    f1 = Fraction(2, 3)
    f2 = Fraction(4, 5)

    print("First Fraction:", f1)
    print("Second Fraction:", f2)

    print("Addition:", f1 + f2)
    print("Subtraction:", f1 - f2)
    print("Multiplication:", f1 * f2)
    print("Division:", f1 / f2)
    print("Decimal value of f1:", f1.convert_to_decimal())
```
