### Encspsulation 
- Encapsulation in Python is an Object-Oriented Programming (OOP) concept where data (variables) and methods (functions) are wrapped together inside a class, and direct access to some parts is restricted.

👉 In simple words:
Encapsulation = Data hiding + controlled access

### 🔹 Why use Encapsulation?
- Protect data from accidental changes
- Improve security
- Make code modular and clean
- Control how data is accessed or modified

Encapsulation is the consept of wraping the data (variable ) and method (function ) together a single unit . it restrict direct access to some of the object component , which is means of preventing accidental interference and misuse of the data.

### 🔹 What is a Private Variable in Python?

A private variable is a variable that cannot be accessed directly from outside the class.

👉 It is defined using double underscore (__)

```python
class Person:
    def __init__(self):
        self.__age = 21   # Private variable

p = Person()

# ❌ This will give error
print(p.__age)
```

### 🔹 Why can't we access it?

Because Python does name mangling internally.

👉 __age becomes:

_Person__age

So direct access is blocked to protect data.

```python
class Person:
    def __init__(self):
        self.__age = 21

    def get_age(self):
        return self.__age

p = Person()
print(p.get_age())   # ✅ Works
```

```python
class Person:
    def __init__(self, name, age):
        self.__name = name   # private variable
        self.__age = age     # private variable

    # Getter for name
    def get_name(self):
        return self.__name

    # Setter for name
    def set_name(self, name):
        self.__name = name

    # Getter for age
    def get_age(self):
        return self.__age

    # Setter for age
    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Age cannot be negative")
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name          # Public variable
        self.__marks = marks      # Private variable

    # Getter method
    def get_marks(self):
        return self.__marks

    # Setter method
    def set_marks(self, marks):
        if marks >= 0 and marks <= 100:
            self.__marks = marks
        else:
            print("Invalid marks")

```

```python
# Object creation
s = Student("Harsh", 85)

# Access using getter
print("Marks:", s.get_marks())

# Modify using setter
s.set_marks(90)
print("Updated Marks:", s.get_marks())

# Invalid update
s.set_marks(150)
```
