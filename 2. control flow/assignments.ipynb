### Assignment 8: break Statement

Write a program that asks the user to input numbers until they input 0. The program should print the sum of all the input numbers.

```python
total = 0

while True:
    num = int(input("enter a number (0 to atop):"))
    if num==0:
        break
    total+=1

print("sum of the entered number is:", total)    
```

### Assignment 12: Factorial Calculation

Write a program that calculates the factorial of a number input by the user using a while loop.



```python
n = int(input("Enter a non-negative integer: "))

# Check for negative input
if n < 0:
    print("Factorial is not defined for negative numbers.")
else:
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    print(f"The factorial of {n} is {fact}")
```


### Assignment 13: Sum of Digits

Write a program that calculates the sum of the digits of a number input by the user using a while loop.

```python
num = int(input("Enter a non-negative integer: "))
original = num
sum_of_digits = 0

while num > 0:
    digit = num % 10         # get the last digit
    sum_of_digits += digit   # add it to sum
    num = num // 10           # remove the last digit

print(f"The sum of the digits of {original} is {sum_of_digits}")
```

### Assignment 14: Prime Number Check

Write a program that checks if a number input by the user is a prime number using a for loop.


```python
num = int(input("Enter a number: "))

if num <= 1:
    print(f"{num} is not a prime number.")
else:
    for i in range(2, num):
        if num % i == 0:
            print(f"{num} is divisible by {i}, so it's not a prime number.")
            break
    else:
        print(f"{num} is a prime number.")
```

### Assignment 15: Fibonacci Sequence

Write a program that prints the first n Fibonacci numbers, where n is input by the user.

```python
n = int(input("Enter how many terms of Fibonacci sequence you want: "))

a, b = 0, 1
print("Fibonacci sequence:")

for _ in range(n):
    print(a, end=" ")
    a, b = b, a + b
```
