```python
## loop control statement 

# break
# the break statement exist the loop prematurely 

# break statement 

for i in range(10):
    if i ==5:
        break
    print(i)
```

```python
a=1
for i in range(a,11):
    if i==7:
        break
    print(i)
```

```python
## continue
## the continue statement the skip the current iteration and continues with the next.
 

for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
## passn
## the pass statement is a null operation; it does nothing

for i in range (5):
    if i ==3:
        print("the number is",i)
        pass
    print(i)
```

### Nested loop

```python
for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

### break & continue in while loop
Break = use to terminate the loop when encountered.
Continue = terminate execution in the current iteration & continue execution of the loop while the next iteration.

```python
## break
i = 1 
while i <=5:
    print(i)
    if(i == 3):
       break
  
    i+=1
print("end of loop")    
```

```python
# searching for a number X in this tuple using loop:
nums = (1, 4, 9, 16, 25, 36, 49, 64, 81, 100)
x =36
i = 0 #initialization
while i <len(nums):
    if(nums[i] == x):
        print("FOUND AT IDX", i )
        break
    else:
        print("finding")
    i+=1
print("end of loop")    
```
