# Python Control Statements (Notes)

- **1. Introduction to Control Statements**  
  - Allow programs to make decisions and control the flow of execution based on conditions.  
  - Execute specific blocks of code only when certain conditions are met.  

- **2. If Statement**  
  - Executes a block of code **only if a condition is True**.  
  - Syntax:  
    ```python
    if condition:
        # code to execute
    ```  
  - Example:  
    ```python
    age = 20
    if age >= 18:
        print("You are an adult.")
    ```

- **3. Else Statement**  
  - Executes a block of code **if the if condition is False**.  
  - Syntax:  
    ```python
    if condition:
        # code if True
    else:
        # code if False
    ```  
  - Example:  
    ```python
    age = 16
    if age >= 18:
        print("You are an adult.")
    else:
        print("You are not an adult.")
    ```

- **4. Elif Statement**  
  - Checks **multiple conditions** after an if statement.  
  - Syntax:  
    ```python
    if condition1:
        # code1
    elif condition2:
        # code2
    else:
        # code3
    ```  
  - Example:  
    ```python
    marks = 80
    if marks >= 90:
        print("Grade: A")
    elif marks >= 75:
        print("Grade: B")
    elif marks >= 50:
        print("Grade: C")
    else:
        print("Grade: D")
    ```

- **5. Nested Conditional Statement**  
  - An `if` or `else` inside another `if` or `else`.  
  - Example:  
    ```python
    age = 20
    has_id = True

    if age >= 18:
        if has_id:
            print("You can vote.")
        else:
            print("You need an ID to vote.")
    else:
        print("You are not eligible to vote.")
    ```

- **6. Practical Example**  
  - Check eligibility for a driving license:  
    ```python
    age = int(input("Enter your age: "))
    citizen = True

    if age >= 18:
        if citizen:
            print("You are eligible for a driving license.")
        else:
            print("You must be a citizen to apply.")
    else:
        print("You are not eligible yet.")
    ```

- **7. Common Errors and Best Practices**  
  - **Common Errors:**  
    - Indentation errors:  
      ```python
      if age >= 18:
      print("Adult")  # ❌
      ```  
      Correct:  
      ```python
      if age >= 18:
          print("Adult")  # ✅
      ```  
    - Using `=` instead of `==` in conditions:  
      ```python
      if age = 18:  # ❌
      if age == 18: # ✅
      ```  
  - **Best Practices:**  
    - Keep conditions simple and readable.  
    - Avoid deeply nested `if` statements.  
    - Use descriptive variable names.
