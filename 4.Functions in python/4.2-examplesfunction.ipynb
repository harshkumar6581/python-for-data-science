## celsius to farenhite
- f =9/5*c  +32

- f = 9/5 *20 +32

### farenhite to celsius 
- c = (f -32)*5/9
- lets suppose we have 77 degree farenhite 
- (77-32)*5/9
- 45*5/9
- 25 c

```python
def convert_temperature(temp,unit):
    """This function converts temperature between Celsius and fahrenheite"""
    if unit=='c':
        return temp *9/5+35  ## celsius to farenhit

    elif unit == "f":
        return(temp-32)*5/9 ## farenhite to celsius
        
    else:
        return None

print(convert_temperature(77,'f'))
print(convert_temperature(25,'c'))          
```

Example 3: Calculate the total cost of items in a shoping cart 

```python
def calculate_total_cost(cart):
    total_cost = 0
    for item in cart:
        # Convert quantity to int to handle string inputs like '4'
        quantity = int(item['quantity'])
        total_cost += item['price'] * quantity
    return total_cost

# Example cart data
cart = [
    {'name': 'Apple', 'price': 0.5, 'quantity': '4'},
    {'name': 'banana', 'price': 0.3, 'quantity': 6},
    {'name': 'orange', 'price': 0.5, 'quantity': 3},
]

# Calling the function
total_cost = calculate_total_cost(cart)
print(total_cost)  ## output is 5.3
```

### Example 4: Check if a string is a palindrome


```python
def is_palindrom(s):
    s=s.lower().replace(" ","")
    return s==s[::-1]
print(is_palindrom("a man a plan a canal panama"))
print(is_palindrom("hello"))

```

#### example 5: calculate the factorials of a number using recursion 


- n!=n*(n-)

```python
def factorial(n):
    if n==0:
        return 1
    else:
        return n *factorial(n-1)

print(factorial(3))        
```

```python
def cal_factorial(n):
    if n==0:
        return 1
    else:
        return n*cal_factorial(n-1) 


print(cal_factorial(5))           
```
