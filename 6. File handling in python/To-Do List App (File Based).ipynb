#### 📝 To-Do List App (File Based)

📌 Description

This is a simple Python To-Do List application that stores tasks in a JSON file.
Users can add tasks, view tasks, mark tasks as complete, and delete tasks.
The program automatically saves data so tasks remain even after closing the program

```python
import json
import os

FILE = "tasks.json"

# load tasks
def load():
    if os.path.exists(FILE):
        with open(FILE, "r") as f:
            return json.load(f)
    return []

# save tasks
def save(tasks):
    with open(FILE, "w") as f:
        json.dump(tasks, f, indent=2)

# add task
def add():
    task = input("Enter task: ")
    tasks = load()
    tasks.append({"task": task, "done": False})
    save(tasks)

# view tasks
def view():
    tasks = load()
    if not tasks:
        print("No tasks\n")
        return
    for i,t in enumerate(tasks,1):
        status = "✔" if t["done"] else "✘"
        print(f"{i}. {t['task']} [{status}]")

# mark complete
def done():
    view()
    tasks = load()
    n = int(input("Task number: ")) - 1
    if 0 <= n < len(tasks):
        tasks[n]["done"] = True
        save(tasks)

# delete task
def delete():
    view()
    tasks = load()
    n = int(input("Task number: ")) - 1
    if 0 <= n < len(tasks):
        tasks.pop(n)
        save(tasks)

# menu
while True:
    print("\n1.Add 2.View 3.Done 4.Delete 5.Exit")
    ch = input("Choice: ")

    if ch=="1": add()
    elif ch=="2": view()
    elif ch=="3": done()
    elif ch=="4": delete()
    elif ch=="5": break

```
